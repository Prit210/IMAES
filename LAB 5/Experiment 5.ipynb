Knowledge Base

The knowledge base below models common car-starting faults, including the **Starter Motor Fault** (Exercise 1).

```python

rules = {
    "Battery Fault": [
        ["dashboard lights are dim", "engine does not crank"]
    ],
    "Fuel Fault": [
        ["engine cranks but does not start", "fuel gauge shows empty"]
    ],
    "Ignition Fault": [
        ["engine cranks but does not start", "no spark at spark plug"]
    ],


    "Starter Motor Fault": [
        ["clicking sound heard when starting", "engine does not crank"]
    ]
}

hypothesis_order = ["Battery Fault", "Fuel Fault", "Ignition Fault", "Starter Motor Fault"]

print("Knowledge base loaded with hypotheses:", list(rules.keys()))
```

Working Memory and WHY-Trace

```python
known_facts = {}   
rule_trace = {}    # fact -> (hypothesis being tested, rule/conditions list) for WHY explanations

def reset_engine():
    """Clears working memory so a fresh diagnosis session can be run."""
    known_facts.clear()
    rule_trace.clear()
    print("Engine reset. All known facts cleared.")
```

 Askable Facts (with WHY option) — Exercise 3

```python
def ask_fact(fact, hypothesis, rule):

    rule_trace[fact] = (hypothesis, rule)
    while True:
        ans = input(f"Q: Is it true that '{fact}'? (yes/no/why): ").strip().lower()
        if ans in ("why", "w"):
            print(f"  WHY: I am trying to prove the hypothesis '{hypothesis}'.")
            print(f"       This requires the rule: IF {' AND '.join(rule)} THEN {hypothesis}.")
            print(f"       '{fact}' is a condition of that rule that is not yet known, so I must ask you.\n")
            continue
        elif ans in ("yes", "y"):
            known_facts[fact] = True
            return True
        elif ans in ("no", "n"):
            known_facts[fact] = False
            return False
        else:
            print("  Please answer 'yes', 'no', or 'why'.")
```

Backward Chaining Engine

```python
def evaluate_condition(cond, hypothesis, rule):

    if cond in known_facts:
        return known_facts[cond]
    if cond in rules:
        return backward_chain(cond)
    return ask_fact(cond, hypothesis, rule)


def backward_chain(hypothesis):

    if hypothesis in known_facts:
        return known_facts[hypothesis]

    for rule in rules.get(hypothesis, []):
        if all(evaluate_condition(cond, hypothesis, rule) for cond in rule):
            known_facts[hypothesis] = True
            return True

    known_facts[hypothesis] = False
    return False


def diagnose(order):
    
    print("=== Car Fault Diagnosis (Backward Chaining) ===")
    for h in order:
        print(f"\n Testing hypothesis: {h} ")
        if backward_chain(h):
            print(f"\n Diagnosis result: {h} is CONFIRMED.")
            return h
    print("\nNo fault could be confirmed from the given hypotheses.")
    return None
```

# Exercise 1: Diagnose with Starter Motor Fault included

```python
reset_engine()

result = diagnose(hypothesis_order)
```
